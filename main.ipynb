# Análise de Dados

## 1. Análise do faturamento

```python
print('Faturamento total - Loja 1: ', df_loja1['Preço'].sum())
print('Faturamento total - Loja 2: ', df_loja2['Preço'].sum())
print('Faturamento total - Loja 3: ', df_loja3['Preço'].sum())
print('Faturamento total - Loja 4: ', df_loja4['Preço'].sum())

faturamentos = {
    "Loja 1": df_loja1["Preço"].sum(),
    "Loja 2": df_loja2["Preço"].sum(),
    "Loja 3": df_loja3["Preço"].sum(),
    "Loja 4": df_loja4["Preço"].sum(),
}

pd.Series(faturamentos).plot(kind="barh", title="Faturamento por Loja")
```

## 2. Vendas por Categoria

```python
import matplotlib.pyplot as plt

# === Preparar os dados ===
dados1 = df_loja1["Categoria do Produto"].value_counts()
dados2 = df_loja2["Categoria do Produto"].value_counts()
dados3 = df_loja3["Categoria do Produto"].value_counts()
dados4 = df_loja4["Categoria do Produto"].value_counts()

# === Criar a figura 2x2 ===
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# --- Loja 1 ---
axes[0, 0].barh(dados1.index, dados1.values)
axes[0, 0].set_title("Categorias mais vendidas – Loja 1")
axes[0, 0].set_xlabel("Quantidade de Produtos Vendidos")
axes[0, 0].set_ylabel("Categoria")
for i, v in enumerate(dados1.values):
    axes[0, 0].text(v, i, str(v), va="center")

# --- Loja 2 ---
axes[0, 1].barh(dados2.index, dados2.values)
axes[0, 1].set_title("Categorias mais vendidas – Loja 2")
axes[0, 1].set_xlabel("Quantidade de Produtos Vendidos")
axes[0, 1].set_ylabel("Categoria")
for i, v in enumerate(dados2.values):
    axes[0, 1].text(v, i, str(v), va="center")

# --- Loja 3 ---
axes[1, 0].barh(dados3.index, dados3.values)
axes[1, 0].set_title("Categorias mais vendidas – Loja 3")
axes[1, 0].set_xlabel("Quantidade de Produtos Vendidos")
axes[1, 0].set_ylabel("Categoria")
for i, v in enumerate(dados3.values):
    axes[1, 0].text(v, i, str(v), va="center")

# --- Loja 4 ---
axes[1, 1].barh(dados4.index, dados4.values)
axes[1, 1].set_title("Categorias mais vendidas – Loja 4")
axes[1, 1].set_xlabel("Quantidade de Produtos Vendidos")
axes[1, 1].set_ylabel("Categoria")
for i, v in enumerate(dados4.values):
    axes[1, 1].text(v, i, str(v), va="center")

plt.tight_layout()
plt.show()
```


## 3. Média de Avaliação das Lojas

```python
# Calcular as médias de avaliação
medias = {
    "Loja 1": df_loja1["Avaliação da compra"].mean(),
    "Loja 2": df_loja2["Avaliação da compra"].mean(),
    "Loja 3": df_loja3["Avaliação da compra"].mean(),
    "Loja 4": df_loja4["Avaliação da compra"].mean(),
}

lojas = list(medias.keys())
valores = list(medias.values())

plt.figure(figsize=(8, 5))
plt.bar(lojas, valores)

plt.title("Média de Avaliação por Loja")
plt.ylabel("Média das Avaliações")

# Inserir valores acima das barras
for i, v in enumerate(valores):
    plt.text(i, v + 0.02, f"{v:.2f}", ha="center")

plt.ylim(0, max(valores) + 0.5)
plt.tight_layout()
plt.show()
```

## 4. Produtos Mais e Menos Vendidos

```python
# Loja 1
vendas1 = df_loja1["Produto"].value_counts()
mais_vendido_1 = vendas1.idxmax()
menos_vendido_1 = vendas1.idxmin()

# Loja 2
vendas2 = df_loja2["Produto"].value_counts()
mais_vendido_2 = vendas2.idxmax()
menos_vendido_2 = vendas2.idxmin()

# Loja 3
vendas3 = df_loja3["Produto"].value_counts()
mais_vendido_3 = vendas3.idxmax()
menos_vendido_3 = vendas3.idxmin()

# Loja 4
vendas4 = df_loja4["Produto"].value_counts()
mais_vendido_4 = vendas4.idxmax()
menos_vendido_4 = vendas4.idxmin()

print("Loja 1 → Mais vendido:", mais_vendido_1, "| Menos vendido:", menos_vendido_1)
print("Loja 2 → Mais vendido:", mais_vendido_2, "| Menos vendido:", menos_vendido_2)
print("Loja 3 → Mais vendido:", mais_vendido_3, "| Menos vendido:", menos_vendido_3)
print("Loja 4 → Mais vendido:", mais_vendido_4, "| Menos vendido:", menos_vendido_4)
```

## 5. Frete Médio por Loja

```python
media_frete_loja1 = df_loja1["Frete"].mean()
media_frete_loja2 = df_loja2["Frete"].mean()
media_frete_loja3 = df_loja3["Frete"].mean()
media_frete_loja4 = df_loja4["Frete"].mean()

print(f"Custo médio de frete - Loja 1: {media_frete_loja1:.2f}")
print(f"Custo médio de frete - Loja 2: {media_frete_loja2:.2f}")
print(f"Custo médio de frete - Loja 3: {media_frete_loja3:.2f}", )
print(f"Custo médio de frete - Loja 4: {media_frete_loja4:.2f}", )

# Gráfico
lojas = ["Loja 1", "Loja 2", "Loja 3", "Loja 4"]
valores = [
    media_frete_loja1,
    media_frete_loja2,
    media_frete_loja3,
    media_frete_loja4
]

plt.figure(figsize=(8,5))
plt.bar(lojas, valores)

plt.title("Custo Médio de Frete por Loja")
plt.ylabel("Valor Médio do Frete (R$)")

# Colocar valores acima das barras
for i, v in enumerate(valores):
    plt.text(i, v + 0.05, f"R$ {v:.2f}", ha="center")

plt.tight_layout()
plt.show()
```
